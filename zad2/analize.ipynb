```python
df = pd.read_csv("results/experiment_results.csv")
```

```python
group_cols = ["activation", "optimizer", "structure", "dataset"]
mean_df = df.groupby(group_cols).agg({
    "mse_train": "mean",
    "mse_val": "mean",
    "mse_test": "mean",
    "epochs_run": "mean",
    "fit_time_sec": "mean"
}).reset_index()

mean_df
```
